# Human Face Recognition System
This notebook implements the face recognition algorithm using the `DeepFace` library.

### Step 1: Prepare the Face Database
We will create a folder named `my_db` where you should upload images of known people. Each image should be named after the person (e.g., `elon_musk.jpg`).

```python
import os
import matplotlib.pyplot as plt
from deepface import DeepFace

# Create a directory for our face database if it doesn't exist
DB_PATH = "my_db"
if not os.path.exists(DB_PATH):
    os.makedirs(DB_PATH)
    print(f"Directory '{DB_PATH}' created. Please upload training images there.")
else:
    print(f"Directory '{DB_PATH}' already exists.")
```

### Step 2: Implementation of the Recognition Algorithm
We will use `DeepFace.find` which encapsulates the steps of detection, alignment, normalization, feature extraction, and distance comparison (Step 5 to 14 of your algorithm).

```python
def recognize_face(test_img_path, db_path, model_name='VGG-Face', distance_metric='cosine'):
    """
    Implements the HumanFaceRecognition Algorithm logic.
    """
    try:
        # DeepFace.find performs detection, alignment, embedding extraction,
        # and compares against the database using the specified metric and threshold.
        results = DeepFace.find(img_path=test_img_path,
                               db_path=db_path,
                               model_name=model_name,
                               distance_metric=distance_metric,
                               enforce_detection=True)

        # results is a list of dataframes (one for each face detected in test image)
        if len(results) > 0 and not results[0].empty:
            # The first result in the dataframe is the one with the minimum distance (Step 13)
            match = results[0].iloc[0]
            identity = match['identity']
            distance = match[f"{model_name}_{distance_metric}"]

            print(f"Match Found: {identity} with distance {distance:.4f}")

            # Step 15: Display
            img = plt.imread(test_img_path)
            plt.imshow(img)
            plt.title(f"Identity: {os.path.basename(identity)} (Dist: {distance:.2f})")
            plt.axis('off')
            plt.show()
        else:
            print("Output: Unknown Person (No suitable match found below threshold)")

    except ValueError as e:
        print(f"Error: {e}. (Likely no face detected in the test image)")

# Example usage (Uncomment after uploading images to my_db and providing a test image):
# recognize_face(test_img_path='test.jpg', db_path=DB_PATH)
```

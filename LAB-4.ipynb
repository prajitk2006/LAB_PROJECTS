```python
import numpy as np
import matplotlib.pyplot as plt

# ==============================================================
# PART A: MAXIMUM LIKELIHOOD ESTIMATION (One-Sample Problem)
# ==============================================================

def mle_one_sample(data):
    """
    Compute MLE for mean and variance of a one-sample dataset.
    Assumes data is normally distributed.
    """
    mu_mle = np.mean(data)
    sigma_mle = np.sum((data - mu_mle)**2) / len(data)  # MLE for variance (not unbiased)
    return mu_mle, sigma_mle

# Example dataset 1
x1 = np.array([12, 15, 14, 10, 13, 16, 11, 14, 15, 13, 35, 20, 37, 10, 40, 27])
mu, var = mle_one_sample(x1)
print("----- Maximum Likelihood Estimation (Dataset 1) -----")
print(f"MLE of Mean     = {mu:.3f}")
print(f"MLE of Variance = {var:.3f}")

# Example dataset 2 (try your own!)
x2 = np.array([5, 7, 8, 6, 9, 10])
mu, var = mle_one_sample(x2)
print("\n----- Maximum Likelihood Estimation (Dataset 2) -----")
print(f"MLE of Mean     = {mu:.3f}")
print(f"MLE of Variance = {var:.3f}")

# ==============================================================
# PART B: LEAST SQUARES LINEAR REGRESSION
# ==============================================================

def least_squares_regression(x, y):
    """
    Compute slope and intercept for least squares linear regression.
    """
    x_mean = np.mean(x)
    y_mean = np.mean(y)
    b = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean)**2)  # Fixed: added division
    a = y_mean - b * x_mean
    y_pred = a + b * x
    return a, b, y_pred

# Example dataset 1
x_reg1 = np.array([1, 2, 3, 4, 5, 6])
y_reg1 = np.array([2, 4, 5, 4, 6, 7])
a, b, y_pred1 = least_squares_regression(x_reg1, y_reg1)
print("\n----- Least Squares Estimation (Dataset 1) -----")
print(f"Slope     = {b:.3f}")
print(f"Intercept = {a:.3f}")
print(f"Regression Equation: y = {a:.3f} + {b:.3f}x")

# Example dataset 2 (try your own!)
x_reg2 = np.array([10, 20, 30, 40, 50])
y_reg2 = np.array([15, 25, 35, 45, 55])
a, b, y_pred2 = least_squares_regression(x_reg2, y_reg2)
print("\n----- Least Squares Estimation (Dataset 2) -----")
print(f"Slope     = {b:.3f}")
print(f"Intercept = {a:.3f}")
print(f"Regression Equation: y = {a:.3f} + {b:.3f}x")

# ==============================================================
# PLOTTING
# ==============================================================

def plot_regression(x, y, y_pred, title):
    plt.scatter(x, y, label="Observed Data", color='blue')
    plt.plot(x, y_pred, label="Least Squares Line", color='red')
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.show()

plot_regression(x_reg1, y_reg1, y_pred1, "Least Squares Linear Regression (Dataset 1)")
plot_regression(x_reg2, y_reg2, y_pred2, "Least Squares Linear Regression (Dataset 2)")
```
